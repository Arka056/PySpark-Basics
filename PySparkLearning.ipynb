Handling Missing Values using Pyspark

*   By Mean, Median, Mode



```python
df_pyspark=spark.read.csv('test1.csv', header=True, inferSchema=True)
```

```python
## To drop all null valued rows
df_pyspark.na.drop().show()
## or
df_pyspark.na.drop(how='any').show()
```

```python
## To drop any null valued rows->> any == how
df_pyspark.na.drop(how='all').show()
```

```python
### Threshold

df_pyspark.na.drop(how='any', thresh=2).show()
```

```python
## subset
df_pyspark.na.drop(how='any', subset=['Age']).show()

## delete the row with null subset column
```

```python
## Filling the Missing value in space of null
df_pyspark.na.fill('Missing Value', ['Name', 'Age']).show()
```

```python
df_pyspark.show()
```

```python
## Replacing the null values using imputer function i.e using Mean,
## Median and Mode values of the specific rows

from pyspark.ml.feature import Imputer

imputer=Imputer(
    inputCols=['Age', 'Experience', 'Salary'],
    outputCols=["{}_imputed".format(c) for c in ['Age', 'Experience', 'Salary']]
).setStrategy('median')
```

```python
## Add imputation cols to df
df_imputed = imputer.fit(df_pyspark).transform(df_pyspark)
df_imputed.show()
```

**Dataframe API**

## Creating Vector Assembler for group
 [Age, Experience] #----> new feature--->independent feature

```python
from pyspark.ml.feature import VectorAssembler
# Use the imputed columns to avoid null pointer exceptions during assembly
featureassembler = VectorAssembler(inputCols=['Age_imputed', 'Experience_imputed'], outputCol='Independent Features')
```

```python
output = featureassembler.transform(df_imputed)
```

```python
output.columns
```

```python
output.show()
```

```python
finalized_data=output.select('Independent Features', 'Salary_imputed')
```

```python
finalized_data.show()
```

```python
from pyspark.ml.regression import LinearRegression
## train test split
train_data, test_data = finalized_data.randomSplit([0.75, 0.25])  #75% of training and 25% of test
regressor = LinearRegression(featuresCol='Independent Features', labelCol='Salary_imputed')
regressor = regressor.fit(train_data)
```

```python
## Coefficients
regressor.coefficients
```

```python
## Intercepts
regressor.intercept
```

```python
## Prediction
pred_results = regressor.evaluate(test_data)
pred_results.predictions.show()
```

```python
pred_results.meanAbsoluteError, pred_results.meanSquaredError
```
